```python
from tensorflow.keras.utils import load_img, img_to_array, array_to_img
import tensorflow as tf
import matplotlib.pyplot as plt

# Load the most beautiful picture in the world
img = load_img('./data/benfica.jpg')
print('Original')
plt.imshow(img)
plt.show()

# Create augmentation model using Keras preprocessing layers (ImageDataGenerator was removed in Keras 3)
augmentation_model = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(15 / 360),           # rotation_range=15 degrees
    tf.keras.layers.RandomTranslation(0.1, 0.1),        # width/height_shift_range=0.1
    tf.keras.layers.RandomZoom((-0.2, 0.2)),             # zoom_range=0.2
    tf.keras.layers.RandomFlip("horizontal"),            # horizontal_flip=True
])

# Preprocess data
X = img_to_array(img)
X = tf.expand_dims(X, 0)  # Add batch dimension

# Apply transformation
print('Transformed')
for i in range(3):  # Generate three transformed pictures
    augmented = augmentation_model(X, training=True)
    plt.imshow(array_to_img(augmented[0]))
    plt.show()
```
